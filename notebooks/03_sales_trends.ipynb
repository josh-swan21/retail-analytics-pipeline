```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

from setup.load_customers import load_customers
df = load_customers()
df.describe()
```

```python
total_spent_by_state = df.groupby('state').agg({'total_spent': 'sum'}).sort_values('total_spent', ascending=False)
top_10_states = total_spent_by_state.head(10)
print(top_10_states)
```

```python
total_cust_by_state =df.groupby('state').agg({'customer_id': 'nunique'}).sort_values('customer_id', ascending=False)
top_10_states_cust = total_cust_by_state.head(10)
print(top_10_states_cust)
```
